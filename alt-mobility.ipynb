```python
import pandas as pd

# Load the CSV
payments_df = pd.read_csv("/kaggle/input/alt-mobility-assessment/payments.csv")

# Drop columns you don't need
payments_df = payments_df.drop(columns=["Unnamed: 6", "Unnamed: 7"])
```

```python
# For payments.csv
payments_df["payment_date"] = pd.to_datetime(payments_df["payment_date"], dayfirst=True, errors="coerce")

# For customer_orders.csv
customer_orders_df = pd.read_csv("/kaggle/input/alt-mobility-assessment/customer_orders.csv")
customer_orders_df["order_date"] = pd.to_datetime(customer_orders_df["order_date"], dayfirst=True, errors="coerce")
```

```python
print(payments_df.isnull().sum())
print(customer_orders_df.isnull().sum())
```

```python
print("\nDuplicate rows in payments:", payments_df.duplicated().sum())
print("Duplicate rows in customer orders:", customer_orders_df.duplicated().sum())
```

```python
print("\nPayments Summary:")
print(payments_df.describe(include='all'))
print("\nPayment Date Range:", payments_df["payment_date"].min(), "to", payments_df["payment_date"].max())
```

```python
import warnings
warnings.filterwarnings('ignore', category=RuntimeWarning)
print("\nCustomer Orders Summary:")
print(customer_orders_df.describe(include='all'))
print("\nOrder Date Range:", customer_orders_df["order_date"].min(), "to", customer_orders_df["order_date"].max())
```

```python
payments_df.to_csv("cleaned_payments.csv", index=False)
customer_orders_df.to_csv("cleaned_customer_orders.csv", index=False)
```
